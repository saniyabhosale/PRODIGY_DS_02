```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('train.csv')

print("--- Initial Data Inspection ---")
print("First 5 rows of the dataset:")
print(df.head())
print("\nInformation about the dataset before cleaning:")
print(df.info())
print("\nMissing values before cleaning:")
print(df.isnull().sum())
```

```python
# --- Data Cleaning ---

# Drop 'Cabin' column due to a large number of missing values
df = df.drop('Cabin', axis=1)
print("\n'Cabin' column dropped.")

# Impute missing 'Age' values with the median
median_age = df['Age'].median()
df['Age'].fillna(median_age, inplace=True)
print(f"Missing 'Age' values imputed with median: {median_age}")

# Impute missing 'Embarked' values with the mode
mode_embarked = df['Embarked'].mode()[0]
df['Embarked'].fillna(mode_embarked, inplace=True)
print(f"Missing 'Embarked' values imputed with mode: {mode_embarked}")

# Check for duplicate rows
num_duplicates_before = df.duplicated().sum()
df.drop_duplicates(inplace=True)
num_duplicates_after = df.duplicated().sum()
print(f"\nNumber of duplicate rows before dropping: {num_duplicates_before}")
print(f"Number of duplicate rows after dropping: {num_duplicates_after}")

print("\n--- Data Information After Cleaning ---")
print(df.info())
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# --- Exploratory Data Analysis (EDA) ---

print("\n--- Descriptive Statistics for Numerical Columns ---")
print(df.describe())

# Univariate analysis - Numerical features
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df['Age'], kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['Age'])
plt.title('Box Plot of Age')
plt.ylabel('Age')
plt.show() # Display plot

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df['Fare'], kde=True)
plt.title('Distribution of Fare')
plt.xlabel('Fare')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['Fare'])
plt.title('Box Plot of Fare')
plt.ylabel('Fare')
plt.show() # Display plot

# Univariate analysis - Categorical features
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
sns.countplot(x='Survived', data=df)
plt.title('Distribution of Survived')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')

plt.subplot(2, 3, 2)
sns.countplot(x='Pclass', data=df)
plt.title('Distribution of Pclass')
plt.xlabel('Passenger Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
plt.ylabel('Count')

plt.subplot(2, 3, 3)
sns.countplot(x='Sex', data=df)
plt.title('Distribution of Sex')
plt.xlabel('Sex')
plt.ylabel('Count')

plt.subplot(2, 3, 4)
sns.countplot(x='SibSp', data=df)
plt.title('Distribution of SibSp')
plt.xlabel('Number of Siblings/Spouses Aboard')
plt.ylabel('Count')

plt.subplot(2, 3, 5)
sns.countplot(x='Parch', data=df)
plt.title('Distribution of Parch')
plt.xlabel('Number of Parents/Children Aboard')
plt.ylabel('Count')

plt.subplot(2, 3, 6)
sns.countplot(x='Embarked', data=df)
plt.title('Distribution of Embarked')
plt.xlabel('Port of Embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)')
plt.ylabel('Count')

plt.tight_layout()
plt.show() # Display plot

# Bivariate analysis - Survived vs. Categorical features
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
sns.barplot(x='Pclass', y='Survived', data=df, ci=None)
plt.title('Survival Rate by Pclass')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')

plt.subplot(2, 3, 2)
sns.barplot(x='Sex', y='Survived', data=df, ci=None)
plt.title('Survival Rate by Sex')
plt.xlabel('Sex')
plt.ylabel('Survival Rate')

plt.subplot(2, 3, 3)
sns.barplot(x='SibSp', y='Survived', data=df, ci=None)
plt.title('Survival Rate by SibSp')
plt.xlabel('Number of Siblings/Spouses Aboard')
plt.ylabel('Survival Rate')

plt.subplot(2, 3, 4)
sns.barplot(x='Parch', y='Survived', data=df, ci=None)
plt.title('Survival Rate by Parch')
plt.xlabel('Number of Parents/Children Aboard')
plt.ylabel('Survival Rate')

plt.subplot(2, 3, 5)
sns.barplot(x='Embarked', y='Survived', data=df, ci=None)
plt.title('Survival Rate by Embarked')
plt.xlabel('Port of Embarkation')
plt.ylabel('Survival Rate')

plt.tight_layout()
plt.show() # Display plot

# Bivariate analysis - Survived vs. Numerical features
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.violinplot(x='Survived', y='Age', data=df)
plt.title('Survival by Age')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')

plt.subplot(1, 2, 2)
sns.violinplot(x='Survived', y='Fare', data=df)
plt.title('Survival by Fare')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')

plt.tight_layout()
plt.show() # Display plot

# Correlation analysis
numerical_cols = df.select_dtypes(include=np.number).columns
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show() # Display plot

print("\nData cleaning and EDA completed. Plots displayed sequentially.")
```
